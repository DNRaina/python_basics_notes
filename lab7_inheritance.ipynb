```python
class Employee:
    def __init__(self, ename, salary):
        self._ename = ename      # Leading underscore signals "protected / internal use"
        self._salary = salary

    # --- GETTER ---
    @property
    def salary(self):
        """Getter: Called automatically when you do `emp.salary`"""
        return self._salary

    # --- SETTER ---
    @salary.setter
    def salary(self, value):
        """Setter: Called automatically when you do `emp.salary = 40000`"""
        if value < 0:
            raise ValueError("Salary cannot be negative!")
        self._salary = value

# Usage:
emp = Employee("Bunty", 30000)

print(emp.salary)       # Calls the getter -> 30000
emp.salary = 45000      # Calls the setter -> updates salary
# emp.salary = -100     # Raises ValueError: Salary cannot be negative!
```

```python
class Grandparent:
    pass

class Parent(Grandparent):
    pass

class Child(Parent):
    pass
```

```python
class Parent:
    def __init__(self, name):
        self.name = name

class Child(Parent):
    def __init__(self, name, age):
        super().__init__(name)  # Calls Parent's __init__ to set name
        self.age = age
```

```python
class Employee:
    company = "Google"

    @classmethod
    def change_company(cls, new_name):
        cls.company = new_name

Employee.change_company("Alphabet")
print(Employee.company)  # Alphabet
```

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Overloading the + operator: v1 + v2
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    # Custom representation when printing: print(v1)
    def __str__(self):
        return f"Vector({self.x}, {self.y})"

v1 = Vector(2, 3)
v2 = Vector(5, 7)
v3 = v1 + v2           # Internally calls v1.__add__(v2)

print(v3)              # Vector(7, 10)
```
