```python
"""types of functions in python
1. built - in functions - already present in python
2. user defined functions - defined by the user"""

# params - placeholder vairables used in a function 
def func(name = "hello ji", uncle = "chache"): #default value
    print("name is", name)
    print("uncle is ", uncle)

func(uncle = "dhruv")
func()
func("name")
```

```python
#functions - args vs kwargs

def fun(*args):
    print(type(args))
    return sum(args)

fun(1,2,3,4)
fun(1,2)
fun()



def func(**kwargs):
    for key, value in kwargs.items():
        print(key, value)

func(name = "alice", age = 30 )
```

```python
##recursion - recursion is when a function calls itself
def func(number):
    if number == 0:
        return
    
    func(number - 1)
    print(number)

func(10)
```
