```python
"""modes of opening a file in python
1. r - open for reading
2. w - open for writing
3. a - a for appending
4. + - open for updating
5. rb - will open for read in binary mode
6. rt - will open for read in text mode
"""
```

```python
with open("filename.txt", "w") as r:
    r.write("hello bro")

with open("filename.txt", "r") as r:
    print(r.read())
#lines are overwritten as it was opened in write mode
```

```python
with open("filename.txt", "r+") as file:
    content = file.read()

    updated_content = content.replace("hello", "****")

    file.seek(0)
    # move the pointer to the first position
    file.write(updated_content)

    file.truncate()
    #truncate the file to remove any leftover old data
```
