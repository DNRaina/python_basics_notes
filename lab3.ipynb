```python
### loops in python

#for loop

for i in range(10):
    print(i, end = " ")
    pass
else:
    print(8) # catches the break

print("\n")
a = 5
while a > 0:
    print(a, end = " ")
    a -= 1
else: #this else statement catches if there is a break in the while loop
    print(a)
```
