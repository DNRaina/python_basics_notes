```python
#class attrtibute - belongs to the class rather than a part. object

class Employee:
    company = "google"

    def getSalary(self):
        print(self.salary)

harry = Employee()
harry.company = "youtube" # limited to harry
Employee.company = "Microsoft" #this changes the class attribute
```

```python
Employee().company
# Employee().salary
```

```python
#instance attribute - belongs to object (instance)

harry.salary = "30k"

harry.salary
```

```python
harry.getSalary()
harry.salary
```

```python
harry.salary
```

```python
#static methods - seomtimes we need a function that doesn't use the self param

@staticmethod
def greet():
    print("hello user")

greet()
```

```python
# __init__() constructor

class Employee:
    def __init__(self, name = "bunty"):
        self.name = name
    def getSalary(self):
        print("salary")

harry = Employee("harry")
```

```python
harry.name
```

```python
Employee.__init__.__defaults__ = ("Aman",)
```
