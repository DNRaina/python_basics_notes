```python
## lists and tuples

"""
a list is always mutable
a tuple however is not"""
```

```python
l1 = [4, "harry"]
l1[0]
l1[:3]
```

```python
# list methods

# l1.sort(reverse= True)
l1.reverse()
print(l1)
l1.append("a value")
print(l1)
l1.insert(0, "value")
print(l1)
l1.pop(1)
print(l1)
l1.remove("value")
print(l1)
```
